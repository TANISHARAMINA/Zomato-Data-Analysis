```python

from google.colab import files
uploaded = files.upload()

import zipfile
with zipfile.ZipFile("archive.zip", 'r') as zip_ref:
    zip_ref.extractall()

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("zomato.csv", encoding="latin-1")

df.drop_duplicates(inplace=True)
df.dropna(inplace=True)


plt.hist(df['Aggregate rating'], bins=20, color='skyblue', edgecolor='black')
plt.title("Rating Distribution")
plt.xlabel("Aggregate Rating")
plt.ylabel("Count")
plt.show()


df['Cuisines'].value_counts().head(10).plot(kind='barh', color='orange')
plt.title("Top 10 Cuisines")
plt.xlabel("Count")
plt.ylabel("Cuisines")
plt.show()

df['City'].value_counts().head(10).plot(kind='barh', color='green')
plt.title("Top 10 Cities")
plt.xlabel("Count")
plt.ylabel("City")
plt.show()


df.boxplot(column='Aggregate rating', by='Price range')
plt.title("Price Range vs Rating")
plt.suptitle("")
plt.xlabel("Price Range")
plt.ylabel("Aggregate Rating")
plt.show()


df.boxplot(column='Aggregate rating', by='Has Online delivery')
plt.title("Online Delivery vs Rating")
plt.suptitle("")
plt.xlabel("Has Online Delivery")
plt.ylabel("Aggregate Rating")
plt.show()


plt.scatter(df['Votes'], df['Aggregate rating'], alpha=0.5, color='red')
plt.title("Votes vs Rating")
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")
plt.show()
```
